```python
class UPI_ID:
    def __init__(self,user_id,bank_id):
        self.user_id=user_id
        self.bank_id=bank_id

    def __repr__(self):
        return f'UPI ID:{{{self.user_id}@{self.bank_id}}}' 
    
    def __eq__(self, other):
        return self.user_id == other.user_id and self.bank_id == other.bank_id

def main():
    shub = UPI_ID('14342346234','okhdfc')
    print(shub.__repr__())
    ujwal = UPI_ID('14342346234','okhdfc')
    
    print(shub.__eq__(ujwal))

main()
    
        
        
```

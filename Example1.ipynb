```python
# 3. Range divisible
start_point = 1
end_point = 20
dividend = 4
for i in range(start_point,end_point+1):
    if(i%dividend==0):
        print(i)
```

```python
#  6. Sum of digits in a number
number = int(input("Enter a number"))
sum=0
while(number>0):
    sum+=number%10
    number=number/10
    print(sum)
    print(number)
print(number)
```

```python
#8. Read a number N and n+nn+nnn
import math
input_number=int(input("Enter a number"))
sum= 0
combined_number=0
for i in range(0,3):
    combined_number+=input_number*(math.pow(10,i))
    sum+=combined_number
print(sum)    
```

```python
#9. Reverse a number
input_number=input("Enter a number")
input_number=input_number[::-1]
reversed_number=int(input_number)
print(reversed_number)
```

```python
# 10. Average in a list
input_list = [10,20,30,40,50,60]
sum=0
for i in input_list:
    sum=sum+i
print(sum/len(input_list))
```

```python
# 11. Count number of Digits in a number
input_number=int(input("Enter a number"))
sum=0
while(input_number!=0):
    input_number=input_number//10
    sum+=1
print(sum)
```

```python
# 12. Check number is palindrome
input_number=input("Enter a number")
reversed_number=input_number[::-1]
if(input_number == reversed_number):
    print("Palindrome")
else:
    print("Not a Palindrome")
```
